Hallazgos:

*Min con numeros negativos en el aparatado de cantidad anomalia encontrada
*Maximos fuera de rango valores atipicos

```python
##Verificacion de tipos de datos
dff.info()
```

```python
##Modificacion de tipo DATE al apartado 'fecha' y atrapar valores basura
dff['fecha'] = pd.to_datetime(dff['fecha'], errors='coerce')
```

```python
##Aplicar strftime para generar la vista de formato adecuada en caso de requerirle
dff['fecha_eua'] = dff['fecha'].dt.strftime('%m-%d-%Y')
```

```python
##Revision de conversion
dff[['fecha', 'fecha_eua']].head(10)
```

```python
dff.head(10)
```

```python
##Auditar y detectar anomalias menos a 0
M_Negativos = dff['cantidad'] < 0
print(f"Registros con valores negativos: {M_Negativos.sum()}")
```

```python
#Filtrar el DataFrame para conservar solo los validos < 0
dff = dff[~M_Negativos].copy()
```

```python
dff.head(10)
```

```python
##Neutralizar atipicos 
print("Registro atipico detectado: ")
display(dff[dff['cantidad'] == 9999.9])
```

```python
dff = dff[dff['cantidad'] != 9999.0].copy()
```

```python
dff
```

```python
##Manejo de nulos
print("Conteo de nulos por columna: ")
print(dff.isnull().sum())
```

```python
##Inicialmente limpiamos las fechas ya que son la columna con menos nulos
##Al ser pocos registros realmente podemos eliminarlos para mantener la integridad del DS
dff = dff.dropna(subset=['fecha', 'fecha_eua']).copy()
```

```python
##Validacion del ajuste
print(f"Dimensiones de dff tras limpiar nulos en fechas: {dff.shape}")
```

```python
##En el tema de la columna monto tambien la eliminamos son pocos registros un 0.8%
print("Eliminar nulos de la columna 'monto' ")  ##Mantener integridad financiera
dff = dff.dropna(subset=['monto']).copy()
```

```python
##Imputar nulos en cliente y nombre (preservar la transaccion)
dff['id_cliente'] = dff['id_cliente'].fillna(-1)
dff['nombre'] = dff['nombre'].fillna('Cliente No Registrado')
```

```python
##Validacion
print(f"Dimensiones de dff tras limpiar el monto e imputar clientes: {dff.shape}")
```

```python
print("Conteo de nulos por columna: ")
print(dff.isnull().sum())
```

```python
##ciudad tiene un 25% de la cantidad de datos es un problema a considerar
##Eliminarlos significa perdida de muchos datos, imputarla podria ser pero lo mas
##limpio y optimo seria preparar una bandera de calidad de datos
##Creacion de (Data Quality Flag)
dff['calidad_geo'] = dff['ciudad'].apply(lambda x: 'Validado' if pd.notnull(x) else 'No registrado')
##Imputamos los nulos
dff['ciudad'] = dff['ciudad'].fillna('Sin asignar')
##Validacion
print("Distribucion de la bandera calidad geografica: ")
print(dff['calidad_geo'].value_counts())
print(f"Shape final de la capa Silver tras resolver nulos: {dff.shape}")
```

```python
dff.head(100)
```

```python
dff['ciudad'] = dff['ciudad'].fillna('Sin Asignar').astype(str).str.strip().str.title()
dff['nombre'] = dff['nombre'].fillna('Desconocido').astype(str).str.strip().str.title()
```

```python
dff.head(100)
```

```python
dff.info()
```

```python
dff['nombre'] = dff['nombre'].astype(str).str.strip() ## Corte de espacios I/F
dff['nombre'] = dff['nombre'].str.title() ##Estandarizacion de textos
```

```python
dff.head(100)
```

```python
##Redefinir los nombres de los headers por higiene conceptual
dff = dff.rename(columns={
    'id_transaccion': 'Id_Transaccion',
    'id_cliente': 'Id_Cliente',
    'fecha': 'Fecha',
    'cantidad': 'Cantidad',
    'monto': 'Monto',
    'nombre': 'Nombre',
    'ciudad': 'ciudad',
    'fecha_eua': 'Fecha_Eua'
})
print(dff.columns.tolist())
```

```python
dff.head(10)
```

```python
##Normalizar el monto a maximo 2 decimales
dff['Monto'] = dff['Monto'].round(2)
```

```python
dff.head(10)
```

```python
dff.info()
```

```python
##Conversion de Fecha_Eua
dff['Fecha_Eua'] = pd.to_datetime(dff['Fecha_Eua'],errors='coerce')
```

```python
dff.info()
```

```python
##Control de duplicados basados en ID
##Conteo de duplicados
duplicados_totales = dff.duplicated(subset=['Id_Transaccion']).sum()
print(f'Registros duplicados detectados: {duplicados_totales}')
```

```python
##Limpieza de duplicados
dff = dff.drop_duplicates(subset=['Id_Transaccion'], keep='first')
```

```python
##Validar volumetria final post-limpieza
print(f'Volumetria final de registros limpios: {len(dff)}')
```

```python
##Guardado de dataframe limpio

dff.to_csv(r'C:\NumpyAndPandas\ETLPORT\Capa_Silver_Transacciones.csv', index=False, encoding= 'utf-8-sig')
print("Capa Silver exportar con exito")
```
